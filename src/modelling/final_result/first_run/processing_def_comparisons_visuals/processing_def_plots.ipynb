# Defence chains vs no defence: comparison plots

Runs `processing_def_graphics.plot_all` for every model and saves three plots per model in this folder:

| File | Plot |
|---|---|
| `<model>_pre.png` | the pre-processing chains alone (post = none) |
| `<model>_post.png` | the 15 post-processing chains alone (pre = none) |
| `<model>_combination.png` | each pre-processing chain + each of the 15 post-processing chains |

Each plot has 6 panels: the metric (default: macro F1) on the clean, PGD, Pixle and PGD + Pixle test sets; **Mean Macro (attacked)** = the mean of the metric over the PGD, Pixle and PGD + Pixle sets; and the time per image. Every panel has a gray line at no defence.

Input: `f1_macro.txt` (or another `<metric>.txt`) and `runtime.txt` in `src/modelling/final_final_results_i_promise/`. Pre chains with ConvAE are left out, because ConvAE is not a pre-processing defence. Run this notebook from its own folder.

```python
from pathlib import Path

import matplotlib.pyplot as plt

import processing_def_graphics as G

OUT = Path(".")          # this folder: processing_def_comparisons_visuals/
METRIC = "f1_macro"      # or "accuracy", "recall_macro", ... (any <metric>.txt in the results folder)
MODELS = list(G.TAGS)    # baseline, robust, grl, adversarial_retrained, convae_rn50
print("Results folder:", G.RESULTS)
print("Models:", MODELS)
```

```python
saved = []
for tag in MODELS:
    G.plot_all(tag, out=OUT, metric=METRIC)
    plt.close("all")  # the PNGs are saved; close the figures so the notebook does not keep 15 large figures
    stem = Path(G.TAGS[tag]).stem
    saved += [OUT / f"{stem}_{kind}.png" for kind in ("pre", "post", "combination")]

missing = [p for p in saved if not p.exists()]
print(f"Saved {len(saved) - len(missing)} of {len(saved)} plots to {OUT.resolve()}")
for p in saved:
    print(" ", p.name)
assert not missing, f"Not saved: {missing}"
```
